```python

# Compound Interest Calculator - FinCAT.inc

p = float(input("Enter your principal (₹): "))

t = float(input("Enter time in years: "))

r = float(input("Enter rate of interest (%): "))

a = p * (1 + r/100)**t

ci = a - p

print(f"\n--- FinCAT Results ---")

print(f"Principal Amount:  ₹{p:,.2f}")

print(f"Rate of Interest:  {r}%")

print(f"Time Period:       {t} years")

print(f"Final Amount:      ₹{round(a, 2):,.2f}")

print(f"Compound Interest: ₹{round(ci, 2):,.2f}")
```

```python
import pandas as pd

#expense tracker
print("------ExpenseTracker------")
print("Time to set up your budget and track expenses!")

# Function to get a valid float input with error handling and roasting
def get_float_input(prompt):
    while True:
        try:
            value = float(input(prompt))
            return value
        except ValueError:
            print("Seriously? That's not a number. Try again, this isn't rocket science.")

# Get Budget
budget = get_float_input("Enter your budget here: ")

expenses = {}
print("\nNow, let's add your expenses. Type 'done' when you're finished.")

while True:
    category_name = input("Enter an expense category (or 'done' to finish): ").strip()
    if category_name.lower() == 'done':
        break
    if not category_name: # Handle empty input for category name
        print("Please enter a category name or 'done'.")
        continue

    # Get amount for the current category
    amount_prompt = f"Enter the amount for '{category_name}': "
    expense_amount = get_float_input(amount_prompt)

    # Add to existing category or create a new one
    if category_name in expenses:
        expenses[category_name] += expense_amount
    else:
        expenses[category_name] = expense_amount

# Calculate total expenses
total = sum(expenses.values())

print("\n----Aapka total expense----")
for category, amount in expenses.items():
    print(f"{category} = {amount:,.2f}") # Formatted for better readability

print(f"Your total was : {total:,.2f}")

# Convert expenses to a DataFrame and save to CSV
if expenses:
    expenses_df = pd.DataFrame(list(expenses.items()), columns=['Category', 'Amount'])
    expenses_df.to_csv('expenses.csv', index=False)
    print("\nExpenses saved to expenses.csv")
else:
    print("\nNo expenses were entered, so no file was saved.")

if total > budget:
    print("Poor financial decisions! You've overspent your budget.")
elif total < budget:
    print("Warren Buffet IRL! You're under budget!")
else:
    print("Good thing Bro! You've hit your budget right on the mark.")
```
