```python
import re

#word count checker
str1 = str(input("Enter your any paragraph here: "))

try:
    # Attempt to convert the input to a float to check if it's a number
    float(str1)
    # If successful, it means the entire input is a number (integer or float)
    print("Error: Input is a number (integer or float). Please enter a paragraph of text.")
except ValueError:
    # If a ValueError occurs, the input is not purely a number, so proceed with original logic
    # Identify potential "unknown" or non-standard characters
    unknown_chars_found = False
    for char in str1:
        if not (char.isalnum() or char.isspace() or char in ".,?!;:'\"-()[]{}"):
            unknown_chars_found = True
            break

    if unknown_chars_found:
        print("Where the hell did you get this ?")

    # Extract words using a regex that handles alphanumeric characters and apostrophes
    # This treats words like "don't" as a single word and ignores most punctuation as delimiters.
    words = re.findall(r"[a-zA-Z0-9']+", str1)

    # Count the extracted words
    word_count = len(str1)

    print("Your text has:", word_count, "words")
```
